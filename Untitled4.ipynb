```python
from google.colab import files
import pandas as pd
import zipfile
import os
import glob
import shutil

uploaded = files.upload()

os.makedirs("/content/fake_news_data", exist_ok=True)

for filename in uploaded:
    path = "/content/" + filename

    if filename.lower().endswith(".zip"):
        with zipfile.ZipFile(path, "r") as z:
            z.extractall("/content/fake_news_data")

    elif filename.lower().endswith(".csv"):
        shutil.copy(
            path,
            "/content/fake_news_data/" + os.path.basename(filename)
        )

fake_files = glob.glob(
    "/content/fake_news_data/**/Fake.csv",
    recursive=True
)
real_files = glob.glob(
    "/content/fake_news_data/**/True.csv",
    recursive=True
)

if not fake_files or not real_files:
    raise FileNotFoundError(
        "Upload the Kaggle ZIP or both Fake.csv and True.csv"
    )

fake_df = pd.read_csv(fake_files[0])
real_df = pd.read_csv(real_files[0])

fake_df.columns = fake_df.columns.str.strip().str.lower()
real_df.columns = real_df.columns.str.strip().str.lower()

fake_df["label"] = "FAKE"
real_df["label"] = "REAL"

df = pd.concat([fake_df, real_df], ignore_index=True)

print("Dataset loaded successfully!")
print("Fake articles:", len(fake_df))
print("Real articles:", len(real_df))
print("Total articles:", len(df))

display(df.head())
```

```python
import numpy as np

# Keep title and article body
for column in ["title", "text"]:
    if column not in df.columns:
        df[column] = ""

    df[column] = df[column].fillna("").astype(str)

# Combine title and body
df["content"] = (
    df["title"].str.strip() + " " +
    df["text"].str.strip()
).str.strip()

# Remove empty articles and duplicates
df = df[df["content"].str.len() > 0].copy()
df = df.drop_duplicates(subset=["content"])

# Keep only required columns and shuffle
df = df[["content", "label"]]
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("After preprocessing:", df.shape)
print("\nClass distribution:")
print(df["label"].value_counts())

display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="label",
    order=["FAKE", "REAL"]
)

plt.title("Fake vs Real News Distribution")
plt.xlabel("News Category")
plt.ylabel("Number of Articles")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Input and output
X = df["content"]
y = df["label"]

# Split: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create ML pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        stop_words="english",
        max_features=100000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
        sublinear_tf=True
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# Train the model
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# Predict test labels
y_pred = model.predict(X_test)

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred,
    pos_label="FAKE",
    zero_division=0
)

recall = recall_score(
    y_test, y_pred,
    pos_label="FAKE",
    zero_division=0
)

f1 = f1_score(
    y_test, y_pred,
    pos_label="FAKE",
    zero_division=0
)

# Display results
print("MODEL EVALUATION RESULTS")
print("========================")
print(f"Accuracy:         {accuracy:.4f}")
print(f"Precision (FAKE): {precision:.4f}")
print(f"Recall (FAKE):    {recall:.4f}")
print(f"F1-score (FAKE):  {f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    labels=["FAKE", "REAL"],
    zero_division=0
))
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["FAKE", "REAL"]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["FAKE", "REAL"]
)

disp.plot(
    values_format="d",
    cmap="Blues"
)

plt.title("Fake News Detection - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
metrics_df = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score"
    ],
    "Score": [
        accuracy,
        precision,
        recall,
        f1
    ]
})

plt.figure(figsize=(8, 5))

sns.barplot(
    data=metrics_df,
    x="Metric",
    y="Score"
)

plt.ylim(0, 1)
plt.title("Model Performance Evaluation")
plt.ylabel("Score")
plt.xlabel("Evaluation Metric")

plt.tight_layout()
plt.show()

display(metrics_df)
```

```python
print("SHORT RESULT ANALYSIS")
print("=====================")

print(f"Accuracy: {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall: {recall:.2%}")
print(f"F1-score: {f1:.2%}")

print("\nAnalysis:")

if accuracy >= 0.90:
    print("The model correctly classified most test articles.")
else:
    print("The model may need further improvement.")

if precision > recall:
    print("Precision is higher than recall.")
    print("Some fake news articles may remain undetected.")
else:
    print("Recall is at least as high as precision.")
    print("Check false-positive predictions in the confusion matrix.")

print(
    "\nNote: The model learns dataset patterns. "
    "Its prediction does not prove whether news is factually true."
)
```

```python
import gradio as gr

def predict_news(article):
    article = (article or "").strip()

    if len(article.split()) < 5:
        return (
            "Please enter a longer news article "
            "or headline (at least 5 words).",
            {}
        )

    # Predict the class
    prediction = model.predict([article])[0]

    # Calculate class probabilities
    probabilities = model.predict_proba([article])[0]
    classes = model.named_steps["classifier"].classes_

    confidence = {
        str(label): float(probability)
        for label, probability in zip(classes, probabilities)
    }

    if prediction == "FAKE":
        result = (
            "Prediction: FAKE\n"
            "The text resembles fake-news examples "
            "in the training dataset."
        )
    else:
        result = (
            "Prediction: REAL\n"
            "The text resembles real-news examples "
            "in the training dataset."
        )

    result += (
        "\n\nImportant: This is a machine-learning prediction, "
        "not a verified fact-check."
    )

    return result, confidence


app = gr.Interface(
    fn=predict_news,

    inputs=gr.Textbox(
        lines=8,
        label="Enter News Article",
        placeholder="Paste your news headline and article here..."
    ),

    outputs=[
        gr.Textbox(label="Prediction Result"),
        gr.Label(label="Class Probabilities")
    ],

    title="FAKE NEWS DETECTION SYSTEM",

    description=(
        "Machine Learning project using TF-IDF "
        "and Logistic Regression."
    )
)

app.launch(share=True)
```
